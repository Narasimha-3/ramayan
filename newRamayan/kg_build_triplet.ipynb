```python
def parse_triplets(file_path):
    with open(file_path, 'r') as file:
        lines = [line.strip() for line in file if line.strip()]
    
    result = []
    current_set = []
    inside_block = False
    
    for line in lines:
        if line == "jai_balayya_ew":
            if inside_block:
                result.append(current_set)
                current_set = []
            inside_block = not inside_block
        elif inside_block:
            line = line.strip("()")  # Remove parentheses if present
            triplet = tuple(map(str.strip, line.split(',')))
            current_set.append(triplet)
    
    return result

# Example usage:
file_path = "log.txt"
parsed_data = parse_triplets(file_path)
# print(parsed_data[1])
#check if every element is a valid tuple or not, no empty tuple
# for i in range(len(parsed_data)):
#     for j in range(len(parsed_data[i])):
#         if len(parsed_data[i][j])!=3:
#             print("Error at ",i,j)
#             print(parsed_data[i][j])
#write tuple to a file
with open("output.txt","w") as f:
    for i in range(len(parsed_data)):
        for j in range(len(parsed_data[i])):
            f.write(str(parsed_data[i][j])+"\n")
        f.write("\n")
```
